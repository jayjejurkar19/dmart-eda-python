```python
df.replace(['NULL', 'null', 'N/A', 'NA', ''], np.nan, inplace=True)
```

```python
text_columns = df.select_dtypes(include='object').columns
for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df.isnull().sum()
```

```python
df['Gender'] = df['Gender'].fillna('Unknown')
```

```python
product_names = {
    'P1001': 'Aashirvaad Atta 5kg',
    'P1002': 'India Gate Basmati Rice',
    'P1003': 'Fortune Sunflower Oil',
    'P1004': 'Amul Taaza Milk',
    'P1005': 'Amul Butter',
    'P1006': 'Tata Salt',
    'P1007': 'Maggi Noodles',
    'P2001': 'Surf Excel Detergent',
    'P2002': 'Vim Dishwash Bar',
    'P2003': 'Dove Shampoo',
    'P2004': 'Colgate Toothpaste',
    'P3001': 'Cotton Bedsheet',
    'P3002': 'Steel Water Bottle',
    'P3003': 'Cotton T-shirt'
}

df['Product_Name'] = df['Product_Name'].fillna(df['Product_ID'].map(product_names))
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['City'] = df['City'].str.lower()

city_mapping = {
    'nashik': 'Nashik',
    'nasik': 'Nashik',
    'nsk': 'Nashik',
    'pune': 'Pune',
    'poona': 'Pune',
    'mumbai': 'Mumbai',
    'bombay': 'Mumbai',
    'nagpur': 'Nagpur',
    'thane': 'Thane',
    'ahmedabad': 'Ahmedabad',
    'ahemdabad': 'Ahmedabad',
    'surat': 'Surat',
    'indore': 'Indore'
}

df['City'] = df['City'].map(city_mapping)
```

```python
df['State'] = df['State'].str.lower()

state_mapping = {
    'maharashtra': 'Maharashtra',
    'mh': 'Maharashtra',
    'gujarat': 'Gujarat',
    'gj': 'Gujarat',
    'madhya pradesh': 'Madhya Pradesh',
    'mp': 'Madhya Pradesh',
    'm.p.': 'Madhya Pradesh'
}

df['State'] = df['State'].map(state_mapping)
```

```python
df['Gender'] = df['Gender'].str.lower()

gender_mapping = {
    'male': 'Male',
    'm': 'Male',
    'man': 'Male',
    'female': 'Female',
    'f': 'Female',
    'woman': 'Female',
    'other': 'Other',
    'o': 'Other',
    'unknown': 'Unknown'
}

df['Gender'] = df['Gender'].map(gender_mapping).fillna('Unknown')
```

```python
df['Segment'] = df['Segment'].str.lower()

segment_mapping = {
    'retail': 'Retail',
    'ret.': 'Retail',
    'corporate': 'Corporate',
    'corp.': 'Corporate',
    'corp': 'Corporate',
    'b2b': 'Corporate',
    'wholesale': 'Wholesale',
    'whole sale': 'Wholesale',
    'ws': 'Wholesale'
}

df['Segment'] = df['Segment'].map(segment_mapping)
```

```python
df['Category'] = df['Category'].str.lower()

category_mapping = {
    'foods': 'Foods',
    'food': 'Foods',
    'non-foods': 'Non-Foods',
    'non foods': 'Non-Foods',
    'nf': 'Non-Foods',
    'general merchandise & apparel': 'General Merchandise & Apparel',
    'general merchandise': 'General Merchandise & Apparel',
    'gen. merchandise': 'General Merchandise & Apparel',
    'gma': 'General Merchandise & Apparel'
}

df['Category'] = df['Category'].map(category_mapping)
```

```python
df['Sub_Category'] = df['Sub_Category'].str.lower()

subcategory_mapping = {
    'staples': 'Staples',
    'staple': 'Staples',
    'dairy': 'Dairy',
    'milk products': 'Dairy',
    'packaged foods': 'Packaged Foods',
    'packaged food': 'Packaged Foods',
    'personal care': 'Personal Care',
    'home care': 'Home Care',
    'kitchenware': 'Kitchenware',
    'kitchen ware': 'Kitchenware',
    'apparel': 'Apparel',
    'clothing': 'Apparel',
    'home utility': 'Home Utility'
}

df['Sub_Category'] = df['Sub_Category'].map(subcategory_mapping)
```

```python
df['Payment_Mode'] = df['Payment_Mode'].str.lower()

payment_mapping = {
    'upi': 'UPI',
    'unified payments': 'UPI',
    'credit card': 'Credit Card',
    'cc': 'Credit Card',
    'debit card': 'Debit Card',
    'dc': 'Debit Card',
    'cash': 'Cash',
    'csh': 'Cash',
    'net banking': 'Net Banking',
    'nb': 'Net Banking'
}

df['Payment_Mode'] = df['Payment_Mode'].map(payment_mapping)
```

```python
def clean_phone(phone):
    if pd.isna(phone):
        return np.nan

    phone = str(phone)

    phone = phone.replace(' ', '')
    phone = phone.replace('-', '')
    phone = phone.replace('+', '')

    if phone.startswith('91') and len(phone) == 12:
        phone = phone[2:]

    if len(phone) == 10 and phone.isdigit():
        return phone

    return np.nan


df['Phone'] = df['Phone'].apply(clean_phone)
```

```python
df.loc[df['Quantity'] <= 0, 'Quantity'] = np.nan

df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
```

```python
df.loc[(df['Age'] < 18) | (df['Age'] > 100), 'Age'] = np.nan

df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Total_Amount'] = df['Quantity'] * df['Selling_Price']

df['Total_Amount'] = df['Total_Amount'].round(2)
```
